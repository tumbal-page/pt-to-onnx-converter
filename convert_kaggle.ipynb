```python
from pathlib import Path
import shutil
from huggingface_hub import hf_hub_download
from ultralytics import YOLO
import onnxruntime as ort

REPO = 'PSImera/manga_bubbles_detect'
FILE = 'bubbles_detect.pt'
IMGSZ = 1024
OUT = Path('/kaggle/working')

pt = OUT / FILE
shutil.copy(hf_hub_download(repo_id=REPO, filename=FILE), pt)

model = YOLO(str(pt))
onnx_path = model.export(format='onnx', imgsz=IMGSZ, simplify=True)
print('ONNX:', onnx_path)

s = ort.InferenceSession(str(onnx_path), providers=['CPUExecutionProvider'])
for i in s.get_inputs():
    print('in ', i.name, i.shape, i.type)
for o in s.get_outputs():
    print('out', o.name, o.shape, o.type)
print('classes:', model.names)
```
